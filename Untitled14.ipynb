```python
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.preprocessing import OneHotEncoder
gent_encod =OneHotEncoder(drop='first')
```

```python
gent=gent_encod.fit_transform(dd[['Geography']]).toarray()
geo_encoded_df=pd.DataFrame(gent,columns=gent_encod.get_feature_names_out(['Geography']))
```

```python
data = pd.concat([dd.drop(columns=['Geography']),geo_encoded_df],axis=1)
data.head()
```

```python
data.shape
```

```python
data.info()
```

Necesary EDA

```python
from matplotlib import pyplot as plt
```

```python
exited_count=data['Exited'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(exited_count,
        labels=exited_count.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['lightblue','red'],
        explode=(0,0.1),
        shadow=True)
plt.title('Distribution of Exited customer')
plt.axis('equal')
plt.show()
```

```python
data.corr()
```

```python
data.describe()
```

```python
x=data.drop(columns=['Exited'],axis=1)
y =data[['Exited']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, y_train, y_test =train_test_split(x,y,test_size=0.3, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
clt=DecisionTreeClassifier()
```

```python
clt.fit(X_train,y_train)
```

```python
pred=clt.predict(X_test)
pred
```

```python
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
```

```python
from sklearn.model_selection import cross_val_score
```

```python
accuracy_score(y_test,pred)
```

```python
print(classification_report(y_test,pred))
```

```python
cm =confusion_matrix(y_test,pred)
```
